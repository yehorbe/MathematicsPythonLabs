## Problem 2

In this second problem, you need to apply two transformations in a sequence. There are several ways to solve this. Notice, that you can also solve this by continuing from the your solution to the problem 1.


```python
# Run this code to see the transformed graphics
p2 = lab7.problem2(student_id)
lab7.draw(p0, p2)
```

```python
# Solution for problem 2
T = np.array([
    [1, 0, 7.5],
    [0, 1, -5.0],
    [0, 0, 1]
])

theta = np.deg2rad(90)

T2 = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta), np.cos(theta), 0],
    [0, 0, 1]
])

p1 = T @ T2 @ p0
lab7.draw(p0, p1)
```

## Problem 5

This is the toughest problem. Can you solve this using linear transformations? 

Tips:  

- In your transformation sequence you need to *shear* the object 45 degrees in order to get the squeezed face.
- There is NO scaling used to get this transformation.

```python
# Run this code to see the transformed graphics
p5 = lab7.problem5(student_id)
lab7.draw(p0, p5)
```

```python
# Solution for problem 5
T = np.array([
    [1, 0, 4],
    [0, 1, 5],
    [0, 0, 1]
])

theta1 = np.deg2rad(180)
theta2 = np.deg2rad(45)

T2 = np.array([
    [np.cos(theta1), -np.sin(theta1), 0],
    [np.sin(theta1), np.cos(theta1), 0],
    [0, 0, 1]
])

T4 = np.array([
    [1, np.tan(theta2), 0],
    [0, 1, 0],
    [0, 0, 1]
])

p1 = T @ T2 @ T4 @ p0
lab7.draw(p0, p1)
```
