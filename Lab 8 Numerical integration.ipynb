## Problem 1

Write your own function that integrates along the given axis using the composite trapezoidal rule. The function should take two input parameters:
- `y` is the input array to integrate
- `x` is the array containing the sample points correspondings to the `y` values.

Test the function with the following test codes:
```
>>> A1 = my_trapezoid([1, 2, 3])
>>> print(f'A1 = {A:+7.4f}')
A1 = +4.0000

>>> A2 = my_trapezoid([1, 2, 3], [4, 6, 8])
>>> print(f'A2 = {A2:+7.4f}')
A2 = +8.0000

>>> A3 = my_trapezoid([1, 2, 3], [8, 6, 4])
>>> print(f'A3 = {A3:+7.4f}')
A3 = -8.0000

>>> x = np.linspace(0, 1, num=50)
>>> y = x**2
>>> A4 = my_trapezoid(y, x)
>>> print(f'A4 = {A4:+7.4f}')
A4 = +0.3334
```

````python
# My trapezoid
def my_trapezoid(y, x=None):
    if x is None:
        x = np.arange(len(y))
    return np.trapezoid(y, x)


# Your test codes come here
A1 = my_trapezoid([1, 2, 3])
print(f"A1={A1:+7.4f}")

A2 = my_trapezoid([1, 2, 3], [4,6,8])
print(f"A2={A2:+7.4f}")

A3 = my_trapezoid([1, 2, 3], [8,6,4])
print(f"A3={A3:+7.4f}")

x = np.linspace(0, 1, num=50)
y = x**2
A4 = my_trapezoid(y, x)
print(f"A4={A4:+7.4f}")
````

## Problem 3

Write **your own code** for calculating the cumulative integrate of $y(x)$ using the composite trapezoid rule.

 The function should take two input parameters:
- `y` is the input array to integrate
- `x` is the array containing the sample points correspondings to the `y` values.

Test the function with the following code:
```
>>> x = np.linspace(-2, 2, num=20)
>>> y = x
>>> y_int = my_cumulative_trapezoid(y, x)
>>> plt.plot(x, y_int, 'b-')
>>> plt.grid()
>>> plt.show()
```

````python
# Your results should look like this
````

````python
import matplotlib.pyplot as plt
import numpy as np
import scipy.integrate


def my_cumulative_trapezoid(y, x):
    return scipy.integrate.cumulative_trapezoid(y, x, initial=0)


x = np.linspace(-2, 2, num=20)
y = x
y_int = my_cumulative_trapezoid(y, x)

plt.plot(x, y_int, "b-")
plt.title("Cumulative integral of $y = x$")
plt.grid()
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np

ts = 0.01
t = np.arange(0, 10 + ts, ts)

v = -0.5 * t**2 + 5 * t

p = my_cumulative_trapezoid(v, t)

plt.figure(figsize=(6, 5))

plt.subplot(2, 1, 1)
plt.plot(t, v)
plt.title("Velocity")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(t, p)
plt.title("Position")
plt.grid(True)

plt.tight_layout()
plt.show()
````
